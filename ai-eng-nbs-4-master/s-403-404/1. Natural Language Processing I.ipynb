## Introduction to Natural Language Processing (NLP)

Natural Language Processing (NLP) is a branch of artificial intelligence and machine learning that focuses on enabling computers to understand, interpret, generate, and interact with human language in both written and spoken forms.

Common NLP tasks include:
- Text classification (e.g., spam detection, topic categorization)
- Sentiment analysis (determining whether text expresses positive, negative, or neutral sentiment)
- Part-of-speech tagging (labeling words as nouns, verbs, adjectives, etc.)
- Named Entity Recognition (NER) (identifying names of people, organizations, locations, etc.)
- Information extraction (extracting structured information from unstructured text)
- Translation (translating text between languages)
- Summarization (producing concise summaries of longer documents)
- Question answering (answering questions based on provided text or general knowledge)
- Text generation (generating human-like text, such as chatbots or content creation)
- Speech recognition (converting spoken language into text)

<br>

## Text Preprocessing and Feature Extraction in NLP

Text is unstructured data, meaning it does not naturally exist in a numerical format that machine learning algorithms can process. Before a model can learn from text, the data must be transformed into a structured representation. 

This is typically done through text preprocessing, which cleans and standardizes the text, followed by feature extraction, which converts it into numerical features. These steps form the basis of traditional NLP pipelines, while modern deep learning approaches learn text representations automatically with minimal manual preprocessing. 

Understanding both approaches provides a strong foundation for applying machine learning to text data.



<br>

## Traditional NLP Pipeline: Preprocessing and Feature Extraction

Traditional NLP approaches require text to be cleaned and transformed into numerical features before it can be used to train machine learning models. 

This process consists of two main stages: **text preprocessing**, which standardizes and cleans the text, and **feature extraction**, which converts the processed text into numerical representations that machine learning algorithms can understand.

<br>

Techniques for text preprocessing:
- Tokenization
- Remove punctuation, numbers, and/or special characters
- Remove stopwords
- Stemming & Lemmatization

<br>

Techniques for feature extraction:
- BoW (Bag of Words)
- TF-IDF
- N-grams (typically applied with BoW/TF-IDF)
- Embeddings

## Traditional vs. Modern NLP

NLP has evolved from manually engineered pipelines to deep learning models that automatically learn text representations. While modern approaches dominate most applications today, traditional methods remain valuable due to their simplicity, efficiency, and interpretability.

<br>

![](./images/classic-ml-vs-deep-learning.jpg)

<br>

Traditional NLP (relies on manual feature engineering):
- Preprocess text (e.g., tokenization, stopword removal, stemming/lemmatization).
- Convert text into numerical features using Bag of Words (BoW) or TF-IDF, often with n-grams.
- Train a classical ML model (e.g., Logistic Regression, Naive Bayes, SVM).

Modern NLP (relies on deep learning):
- Minimal manual preprocessing.
- Pretrained models (e.g., LSTMs, Transformers) automatically learn meaningful text representations.
- Fine-tune the model for the target NLP task.


<br>

## Use cases

Traditional NLP techniques (e.g., BoW and TF-IDF) remain widely used in practice, especially for:
- Text classification on small or medium-sized datasets.
- Baseline models.
- Resource-constrained environments.
- Applications where interpretability and speed are important.

However, for tasks such as machine translation, question answering, summarization, and modern chatbots, deep learning models (particularly Transformers) have become the standard.
